## 1. Import Libraries

**Why:** pandas provides the table-reading and inspection tools needed for this notebook.

**What we expect to learn:** This step only prepares the notebook; it does not inspect or change any business data.

```python
import pandas as pd
```

**Meaning:** pandas is available and ready to load the orders table. No data has been read or changed yet.

## 2. Load Dataset

**Why:** We need the orders CSV in a DataFrame so we can inspect its structure.

**What we expect to learn:** This creates an in-memory view of the orders table. It reads only the approved CSV and does not write back to it.

```python
orders_path = "../data/raw/olist/olist_orders_dataset.csv"
orders_df = pd.read_csv(orders_path)
```

**Meaning:** The raw orders table is loaded in memory for read-only inspection. No cleaning or modification has been applied.

## 3. Display Shape

**Why:** Shape gives the number of order records and fields at a glance.

**What we expect to learn:** We expect a `(rows, columns)` pair describing the size of this table.

```python
orders_df.shape
```

**Business meaning:** The table contains 99,441 order records described by 8 fields. This establishes the scale of the order-lifecycle data available for later analysis.

## 4. Display First 5 Rows

**Why:** A small preview helps confirm how one order record is organized without analyzing the full dataset.

**What we expect to learn:** We expect to see identifiers, status, and lifecycle timestamps in their raw form.

```python
orders_df.head()
```

**Business meaning:** Each row represents an order linked to a customer, a current/final status, and timestamps for major fulfilment stages. The preview confirms that the table follows the order journey from purchase to estimated and actual delivery.

## 5. Display Column Names

**Why:** Column names define the business fields available in the orders table.

**What we expect to learn:** We expect identifiers, order status, and timestamps covering the order lifecycle.

```python
orders_df.columns
```

**Business meaning:** The available fields support order identification, customer linkage, status monitoring, and measurement of approval and delivery stages. No sales value or product detail exists in this table.

## 6. Display Data Types

**Why:** Data types show how pandas interpreted each raw column.

**What we expect to learn:** We expect identifiers, status, and unparsed timestamp fields to appear as text because no conversion is allowed in this milestone.

```python
orders_df.dtypes
```

**Business meaning:** All eight fields are currently text. The lifecycle dates cannot yet support time calculations without a later, explicitly approved conversion step; this notebook intentionally leaves them unchanged.

## 7. Check Missing Values

**Why:** Missing-value counts reveal which order-lifecycle fields are not populated for every record.

**What we expect to learn:** We expect identifiers and purchase fields to be complete, while some later lifecycle timestamps may be absent for orders that did not reach those stages.

```python
orders_df.isna().sum()
```

**Business meaning:** Order IDs, customer IDs, statuses, purchases, and estimates are complete. Approval is missing for 160 records, carrier handoff for 1,783, and customer delivery for 2,965. These gaps may reflect orders that were canceled, unavailable, or still in progress, but that explanation must be tested later rather than assumed.

## 8. Count Unique Values for `order_status`

**Why:** Status counts show how orders are distributed across fulfilment stages and outcomes.

**What we expect to learn:** We expect to identify the available status categories and the number of records in each category.

```python
orders_df["order_status"].value_counts(dropna=False)
```

**Business meaning:** The table contains eight statuses and is dominated by delivered orders (96,478). Smaller groups represent shipped, canceled, unavailable, invoiced, processing, created, and approved orders, which are important exceptions or unfinished stages for later investigation.

## 9. Basic Summary Using `describe()`

**Why:** `describe()` provides a consistent structural summary of count, distinct values, most common value, and frequency for the text columns.

**What we expect to learn:** We expect to confirm completeness, identifier uniqueness, status variety, and repeated timestamp values without changing data types.

```python
orders_df.describe(include="all")
```

**Business meaning:** Both order and customer IDs are unique in this table, confirming one order-level customer record per order. Status has eight categories, and delivered is the most frequent. Lower counts in later lifecycle timestamps agree with the missing-value check. Repeated timestamps are possible and do not by themselves indicate duplicate orders.

## 10. Business Observations

- The orders table contains 99,441 unique orders and 99,441 unique order-level customer IDs.
- Most records have a `delivered` status, while a smaller set represents unfinished or unsuccessful order outcomes.
- Purchase and estimated-delivery timestamps are complete, but approval, carrier handoff, and actual-delivery timestamps contain missing values.
- Missing later-stage timestamps should be evaluated together with order status in the next approved EDA step; they should not be filled or removed without understanding the business process.
- All timestamp columns remain raw text. Converting them to datetime would be a separate preparation decision and was intentionally not performed here.
- This table describes order progress but does not contain product, seller, payment, review, customer-location, or monetary detail. No other table was analyzed.

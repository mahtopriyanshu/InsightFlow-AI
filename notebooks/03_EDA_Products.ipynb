## 7. Missing values

Count nulls by field.

```python
df.isna().sum()
```

**Observation:** Missing counts identify incomplete fields without changing them.

## 8. Missing percentage

Measure missingness relative to table size.

```python
(df.isna().mean().mul(100).round(2)).sort_values(ascending=False)
```

**Observation:** Percentages show the relative scale of incomplete fields.

## 9. Duplicate rows

Count complete row duplicates.

```python
df.duplicated().sum()
```

**Observation:** This is the number of repeated full rows; none are removed.

## 10. Duplicate primary keys

Validate the documented candidate key.

```python
primary_key = ['product_id']
df.duplicated(subset=primary_key, keep=False).sum()
```

**Observation:** This counts rows involved in duplicate candidate keys for product_id.

## 11. Unique values

Inspect meaningful category or entity counts.

```python
product_counts = pd.Series({'products': df['product_id'].nunique(), 'nonblank_categories': df['product_category_name'].nunique()})
print(product_counts)
df['product_category_name'].value_counts(dropna=False).head(10)
```

**Observation:** This gives a concise view of important distinct values.

## 12. Basic describe()

Generate a basic summary for all fields.

```python
df.describe(include='all')
```

**Observation:** The summary describes completeness, distinctness, common values, and numeric ranges where applicable.

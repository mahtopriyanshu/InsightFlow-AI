## 1. Import libraries

Import pandas for tabular inspection.

```python
import pandas as pd
```

## 2. Load CSV

Load only this CSV into memory. Timestamp columns are converted on the in-memory copy.

```python
file_path = '../data/raw/olist/olist_order_items_dataset.csv'
df = pd.read_csv(file_path)
for column in ['shipping_limit_date']:
    df[column] = pd.to_datetime(df[column], errors='coerce')
```

**Observation:** The table is loaded; the raw CSV is unchanged.

Count nulls by field.

## 8. Missing percentage

Measure missingness relative to table size.

```python
(df.isna().mean().mul(100).round(2)).sort_values(ascending=False)
```

**Observation:** Percentages show the relative scale of incomplete fields.

## 9. Duplicate rows

Count complete row duplicates.

```python
df.duplicated().sum()
```

**Observation:** This is the number of repeated full rows; none are removed.

## 10. Duplicate primary keys

Validate the documented candidate key.

```python
primary_key = ['order_id', 'order_item_id']
df.duplicated(subset=primary_key, keep=False).sum()
```

**Observation:** This counts rows involved in duplicate candidate keys for order_id, order_item_id.

## 11. Unique values

Inspect meaningful category or entity counts.

```python
pd.Series({'orders_with_items': df['order_id'].nunique(), 'products': df['product_id'].nunique(), 'sellers': df['seller_id'].nunique(), 'largest_item_sequence': df['order_item_id'].max()})
```

**Observation:** This gives a concise view of important distinct values.

## 12. Basic describe()

Generate a basic summary for all fields.
